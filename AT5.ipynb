# Análise Exploratória de Dados - Titanic


# 1. Importando as bibliotecas

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# 2. Carregando o dataset


```python
# Carregando o dataset
train = pd.read_csv('train.csv')

# Exibindo as primeiras linhas
train.head()
```

# 3. Quantidade de linhas e colunas do dataset

```python
# Quantidade de linhas e colunas
linhas, colunas = train.shape

print(f'Quantidade de linhas: {linhas}')
print(f'Quantidade de colunas: {colunas}')
```


# 4. Tipo de dado de cada variável

```python
# Tipo de dado das variáveis
train.dtypes
```

# 5. Resumo estatístico das variáveis

```python
# Resumo estatístico
train.describe()
```


# 6. Tabela de frequência e gráfico da variável "Survived"

```python
# Tabela de frequência
freq_survived = train['Survived'].value_counts()
print(freq_survived)
```

```python
# Gráfico da variável Survived
plt.figure(figsize=(6,4))
sns.countplot(x='Survived', data=train)
plt.title('Distribuição de Sobrevivência')
plt.xlabel('Sobreviveu')
plt.ylabel('Quantidade')
plt.xticks([0,1], ['Não', 'Sim'])
plt.show()
```

# 7. Quantas pessoas sobreviveram? Qual a classe mais frequente?

```python
# Quantidade de sobreviventes
sobreviventes = train['Survived'].sum()
print(f'Quantidade de sobreviventes: {sobreviventes}')
```

```python
# Classe mais frequente
classe_mais_frequente = train['Pclass'].mode()[0]
print(f'Classe mais frequente: {classe_mais_frequente}')
```

* O total de sobreviventes corresponde à soma dos valores da variável `Survived`.
* A função `mode()` identifica a classe mais frequente entre os passageiros.


# 8. Relação entre sexo e sobrevivência

## Tabela de frequência

```python
# Tabela cruzada
freq_sex_survival = pd.crosstab(train['Sex'], train['Survived'])
print(freq_sex_survival)
```


## Gráfico

```python
# Gráfico da relação entre sexo e sobrevivência
plt.figure(figsize=(7,5))
sns.countplot(x='Sex', hue='Survived', data=train)
plt.title('Sobrevivência por Sexo')
plt.xlabel('Sexo')
plt.ylabel('Quantidade')
plt.legend(['Não Sobreviveu', 'Sobreviveu'])
plt.show()
```

O gráfico de barras agrupadas foi escolhido porque permite comparar:

* Homens e mulheres;
* Sobreviventes e não sobreviventes.

A análise mostra que:

* Mulheres tiveram uma taxa de sobrevivência significativamente maior;
* Homens morreram em maior quantidade.


# 9. Distribuição das idades dos passageiros

```python
# Gráfico de distribuição das idades
plt.figure(figsize=(8,5))
sns.histplot(train['Age'].dropna(), bins=30, kde=True)
plt.title('Distribuição das Idades')
plt.xlabel('Idade')
plt.ylabel('Frequência')
plt.show()
```


# 10. Análise da distribuição das idades


A distribuição das idades não é perfeitamente simétrica.

Observa-se:

* Maior concentração de passageiros jovens e adultos;
* Pouca quantidade de idosos;
* Leve assimetria à direita.

Portanto, a maioria dos passageiros era jovem.


# 11. Conclusão Geral

A análise exploratória mostrou que:

* A maioria dos passageiros não sobreviveu;
* A classe mais frequente foi a terceira classe (`Pclass = 3`);
* Mulheres sobreviveram proporcionalmente mais que homens;
* A maioria dos passageiros era jovem.
